```python
import pandas as pd
import numpy as np

# 固定隨機種子，確保數值重複執行時一致
np.random.seed(42)

# 建立 60 筆資料與 7 個欄位
df = pd.DataFrame({
    'student_id': range(1, 61),
    'study_hours': np.round(np.random.uniform(0.5, 4.5, 60), 1),
    'sleep_hours': np.round(np.random.uniform(5.0, 9.5, 60), 1),
    'screen_time_hours': np.round(np.random.uniform(1.0, 7.0, 60), 1),
    'attendance_rate': np.random.randint(60, 100, 60),
    'quiz_score': np.random.randint(50, 100, 60),
    'extra_score': np.random.randint(1, 10, 60)
})

# 驗證資料筆數與欄位數
print("資料形狀：", df.shape)
```

```python
# 1. 指定要顯示的欄位清單
selected_columns = ['student_id', 'study_hours', 'sleep_hours', 'screen_time_hours', 'attendance_rate', 'quiz_score']

# 2. 顯示前 5 筆資料
display(df[selected_columns].head(60))
```

```python
# 1. 驗證 Shape 是否為 (60, 7)
print("Shape 驗證：", df.shape)
assert df.shape == (60, 7), "錯誤：資料形狀不是 (60, 7)"

# 2. 驗證 quiz_score 範圍是否在 0 ~ 100 之間
print("quiz_score 最小值：", df['quiz_score'].min())
print("quiz_score 最大值：", df['quiz_score'].max())
assert (df['quiz_score'].min() >= 0) and (df['quiz_score'].max() <= 100), "錯誤：分數超出 0~100 範圍"

print("所有驗證測試皆通過！")
```
